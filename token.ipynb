```python
!pip install -q openai pydantic
from google.colab import userdata
from openai import OpenAI

client = OpenAI (
    api_key = userdata.get ('GEMINI_API_KEY'),
    base_url = "https://generativelanguage.googleapis.com/v1beta/openai/",
)

MODEL = "gemini-3.1-flash-lite"
```

```python
# cost calculator

PRICES = {
    "small": (0.10, 0.40),
    "frontier": (3.00, 15.00),
}

def monthly_cost(requests, in_tok, out_tok, model):
    p_in, p_out = PRICES [model]
    return requests * (in_tok*p_in + out_tok*p_out)/10000000

for m in PRICES:
    print(f"{m:<9} ${monthly_cost(20_000, 1500, 300, m):,.2f} / month")
```

```python
# temp experiment
for t in [0.0, 1.0]:
    names = []
    for _ in range(5):
        r = client.chat.completions.create(
            model=MODEL, temperature=t,
            messages=[{"role": "user", "content": "Suggest one name for a coffee shop. Reply with the name only."}],
        )
        names.append(r.choices[0].message.content.strip())
    print(t, names)
```
