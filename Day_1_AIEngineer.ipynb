```python
%pip install -q openai pydantic
from google.colab import userdata
from openai import OpenAI

client = OpenAI(
    api_key = userdata.get('GEMINI_API_KEY'),
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/",  # Gemini's OpenAI-compatible endpoint
)
MODEL = "gemini-flash-latest"   # alias for the newest Flash model; see ai.google.dev/gemini-api/docs/models
# Using an OpenAI key instead? Drop base_url, use your OpenAI key, and set MODEL to a current OpenAI model.
```

```python
resp = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": "You are a a 10-year-old."},
        {"role": "user", "content": "Explain next-token prediction in 2 sentences."}
    ],

)
print(resp.choices[0].message.content)
print(resp.usage)

# TODO: loop over 3 different system prompts (e.g. "a pirate", "a CFO", "a 10-year-old")
#       and print each answer. What changes? What stays the same?
```
